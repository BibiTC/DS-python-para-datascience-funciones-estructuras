**3** - Crea una función que reciba una lista como parámetro y convierta todos los valores de la lista a flotantes. La función debe incluir un manejo de error indicando el tipo de error generado y devolver la lista si no ha ocurrido ningún error. Por último, debe tener la cláusula finally para imprimir el texto: 'Fin de la ejecución de la función'.

```python
def convertir_flotante(lista:list) ->list:
    '''
        Convierte todos los valores de una lista a flotantes.

        Args:
            lista: La lista a convertir.

        Returns:
            La lista con todos los valores convertidos a flotantes,
            o la lista original si ocurre un error.
    '''

    try:
        # Intentar convertir cada elemento de la lista a flotante
        lista_flotantes = [float(elemento) for elemento in lista]
        return lista_flotantes  # Devolver la lista convertida
    except ValueError as e:
        # Manejar el error si no se puede convertir a flotante
        print(f"Error: {e}. No se pudo convertir un elemento a flotante.")
        return lista  # Devolver la lista original
    finally:
        # Imprimir mensaje al final de la ejecución
        print("Fin de la ejecución de la función")
```

```python
mi_lista = [1, '2.5', 3, '4.0']
lista_convertida = convertir_flotante(mi_lista)
print(lista_convertida)  # Output: [1.0, 2.5, 3.0, 4.0]
```

```python
mi_lista_con_error = [1, '2.5', 'abc', 4]
lista_convertida_error = convertir_flotante(mi_lista_con_error)
print(lista_convertida_error)  # Output: [1, '2.5', 'abc', 4] and an error message.
```

```python
# Opcion del instructor

def convertir_a_float(lista):
    try:
        nueva_lista = [float(valor) for valor in lista]
        return nueva_lista
    except ValueError as e:
        print(f"Error: {e}")
    finally:
        print("Fin de la ejecución de la función")

# Ejemplo de uso:
lista_original = ['4.5', '2.3', '7.8', '1.2']
lista_convertida = convertir_a_float(lista_original)
print(lista_convertida)
```

**4**- Crea una función que reciba dos listas como parámetros y agrupe los elementos uno a uno de las listas, formando una lista de tuplas de 3 elementos. El primer y segundo elemento de la tupla son los valores en la posición i de las listas y el tercer elemento es la suma de los valores en la posición i de las listas.

La función debe incluir un manejo de error indicando el tipo de error generado y devolver como resultado la lista de tuplas. Si las listas enviadas como parámetro tienen tamaños diferentes, la función debe devolver un IndexError con la frase: 'La cantidad de elementos en cada lista es diferente.'.

Datos para probar la función:

Valores sin error:

```python
lista1 = [4, 6, 7, 9, 10]
lista2 = [-4, 6, 8, 7, 9]
```

Listas con tamaños diferentes:

```python
lista1 = [4, 6, 7, 9, 10, 4]
lista2 = [-4, 6, 8, 7, 9]
```

Listas con valores incoherentes:

```python
lista1 = [4, 6, 7, 9, 'A']
lista2 = [-4, 'E', 8, 7, 9]
```

```python
# Desarrollo ejercicio
def agrupar_elementos(lista1, lista2):
    """
    Agrupa elementos de dos listas en tuplas de 3 elementos.

    Args:
        lista1: La primera lista.
        lista2: La segunda lista.

    Returns:
        Una lista de tuplas, o un mensaje de error si ocurre una excepción.
    """
    try:
        if len(lista1) != len(lista2):
            raise IndexError("La cantidad de elementos en cada lista es diferente.")

        lista_tuplas = [(lista1[i], lista2[i], lista1[i] + lista2[i]) for i in range(len(lista1))]
        return lista_tuplas
    except IndexError as e:
        return str(e)  # Devuelve el mensaje de error como una cadena
    except TypeError as e:
        return f"Error: {e}. No se pueden sumar elementos de tipos incompatibles."
```

```python
# Valores sin error
lista1 = [4, 6, 7, 9, 10]
lista2 = [-4, 6, 8, 7, 9]
resultado = agrupar_elementos(lista1, lista2)
print(resultado)  # Output: [(4, -4, 0), (6, 6, 12), (7, 8, 15), (9, 7, 16), (10, 9, 19)]
```

```python
# Listas con tamaños diferentes
lista1 = [4, 6, 7, 9, 10, 4]
lista2 = [-4, 6, 8, 7, 9]
resultado = agrupar_elementos(lista1, lista2)
print(resultado)  # Output: La cantidad de elementos en cada lista es diferente.

```

```python
# Listas con valores incoherentes
lista1 = [4, 6, 7, 9, 'A']
lista2 = [-4, 'E', 8, 7, 9]
resultado = agrupar_elementos(lista1, lista2)
print(resultado)  # Output: Error: unsupported operand type(s) for +: 'int' and 'str'. No se pueden sumar elementos de tipos incompatibles.
```

```python
def calcular_puntuaciones(respuestas_correctas, pruebas_estudiantes):
    """
    Calcula las puntuaciones de los estudiantes en una prueba.

    Args:
        respuestas_correctas: Una lista con las respuestas correctas.
        pruebas_estudiantes: Una lista de listas, donde cada sublista
                            representa las respuestas de un estudiante.

    Returns:
        Una lista con las puntuaciones de cada estudiante, o un mensaje de error
        si se encuentra una alternativa inválida.
    """
    puntuaciones = []
    for prueba in pruebas_estudiantes:
        puntuacion = 0
        for i in range(len(respuestas_correctas)):
            if prueba[i] not in ('A', 'B', 'C', 'D'):
                raise ValueError(f"La alternativa {prueba[i]} no es una opción de alternativa válida")
            if prueba[i] == respuestas_correctas[i]:
                puntuacion += 1
        puntuaciones.append(puntuacion)
    return puntuaciones

```

```python
lista_tratada = ['Python', 'es', 'un', 'lenguaje', 'de', 'programación', 'poderoso', 'versátil',
                  'y', 'fácil', 'de', 'aprender', 'utilizado', 'en', 'diversos', 'campos', 'desde',
                  'análisis', 'de', 'datos', 'hasta', 'inteligencia', 'artificial']
```

```python
# Ayuda de Gemini

def verificar_puntuacion(lista_palabras):
    """
    Verifica si una lista de palabras contiene signos de puntuación.

    Args:
        lista_palabras: Una lista de palabras.

    Raises:
        ValueError: Si se encuentra un signo de puntuación en alguna palabra.
    """
    for palabra in lista_palabras:
        for caracter in palabra:
            if caracter in [',', '.', '!', '?']:
                raise ValueError(f"El texto presenta puntuaciones en la palabra \"{palabra}\"")
```

```python

# Ejemplo de uso
lista_tratada = ['Python', 'es', 'un', 'lenguaje', 'de', 'programación', 'poderoso', 'versátil',
                  'y', 'fácil', 'de', 'aprender', 'utilizado', 'en', 'diversos', 'campos', 'desde',
                  'análisis', 'de', 'datos', 'hasta', 'inteligencia', 'artificial']
lista_no_tratada = ['Python', 'es', 'un', 'lenguaje', 'de', 'programación', 'poderoso,', 'versátil',
                  'y', 'fácil,', 'de', 'aprender', 'utilizado', 'en', 'diversos', 'campos,', 'desde',
                  'análisis', 'de', 'datos', 'hasta', 'inteligencia', 'artificial!']

try:
    verificar_puntuacion(lista_tratada)
    print("La lista tratada no presenta puntuaciones.")
except ValueError as e:
    print(e)  # No se imprimirá nada, ya que la lista tratada no tiene puntuaciones

try:
    verificar_puntuacion(lista_no_tratada)
    print("La lista no tratada no presenta puntuaciones.")  # No se imprimirá
except ValueError as e:
    print(e)  # Output: El texto presenta puntuaciones en la palabra "poderoso,"
```

```python
# Opción de el Instructor
# Definimos la función:

def evaluar_puntuacion(lista_palabras):
    try:
        for palabra in lista_palabras:
            if any(puntuacion in palabra for puntuacion in [',', '.', '!', '?']):
                raise ValueError(f"El texto presenta puntuaciones en la palabra '{palabra}'.")
    except ValueError as e:
        print(f"Error: {e}")
```

7 - Función de división de columnas y manejo de excepciones:

```python
# Creando la función que recibe las dos listas y la operación a realizar
def divide_columnas(lista_1: list, lista_2: list) -> list:
  # try-except que verifica si es posible calcular la división y lanza una excepción si las listas tienen tamaños diferentes
  # o si hay alguna división por cero en uno de los cálculos entre los números de las listas
  try:
    if len(lista_1) != len(lista_2):  # Verificando si las listas tienen el mismo tamaño, si no, lanza una excepción
      raise ValueError("Las listas deben tener el mismo tamaño")

    # La función zip empareja los elementos de las listas y se genera una lista mediante la división entre las parejas
    resultado = [round(a / b, 2) for a, b in zip(lista_1, lista_2)]
  except ValueError as e:
    print(e)
  except ZeroDivisionError as e:
    print(f"{e}: La 2ª lista no puede tener un valor igual a 0")
  else:
    return resultado
```

```python
#Probando sin excepciones:

# Probando en el ejemplo que no lanza excepciones
presiones = [100, 120, 140, 160, 180]
temperaturas = [20, 25, 30, 35, 40]

divide_columnas(presiones, temperaturas)
```
