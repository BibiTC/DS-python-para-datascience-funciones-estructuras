```python
frutas = ["manzana", "banana", "uva", "pera", "mango", "coco", "sandia", "fresa", "naranja", "maracuya", "kiwi", "cereza"]
```

```python
import random

frutas = ["manzana", "banana", "uva", "pera", "mango", "coco", "sandia", "fresa", "naranja", "maracuya", "kiwi", "cereza"]

# Seleccionar 3 frutas aleatorias sin repetición
frutas_sorpresa = random.sample(frutas, 3)

# Imprimir las frutas seleccionadas
print("Ensalada de frutas sorpresa:")
for fruta in frutas_sorpresa:
    print(fruta)
```

9 - Has recibido un desafío para calcular la raíz cuadrada de una lista de números, identificando cuáles resultan en un número entero. La lista es la siguiente:

```python
numeros = [2, 8, 15, 23, 91, 112, 256]
```

```python
import math

numeros = [2, 8, 15, 23, 91, 112, 256]

print("Número | Raíz cuadrada | ¿Es entero?")
print("-------|----------------|------------")

for numero in numeros:
    raiz_cuadrada = math.sqrt(numero)  # Calcula la raíz cuadrada
    es_entero = raiz_cuadrada.is_integer()  # Verifica si es entero
    print(f"{numero:6} | {raiz_cuadrada:14.2f} | {es_entero}")
```
